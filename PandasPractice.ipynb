```python
D1 = {"ID": [1, 2, 3, 4, 5], "Emp Name": ["Silvie", "Drake", "Lucas", "Mia", "Ethan"], "Department": ["HR", "Finance", "IT", "Marketing", "Sales"], "Post": ["Manager", "Analyst", "Developer", "Executive", "Associate"]}
D2 = {"ID": [6,7,8,9,10,11], "Emp Name": ["Olivia", "Liam", "Emma", "Noah", "Nathan", "Brooke"], "Department": ["HR", "IT", "IT", "Finance","Sales", "HR"]}

pd.merge(pd.DataFrame(D1), pd.DataFrame(D2), on = "ID", how = "outer") 
#merge the two DataFrames based on the common column "ID". The "outer" parameter specifies that all rows from both DataFrames should be included in the result, even if there are no matching values in the "ID" column.
```

```python
D1 = {"ID": [1, 2, 3, 4, 5], "Emp Name": ["Silvie", "Drake", "Lucas", "Mia", "Ethan"], "Department": ["HR", "Finance", "IT", "Marketing", "Sales"], "Post": ["Manager", "Analyst", "Developer", "Executive", "Associate"]}
D2 = {"ID": [6,7,8,9,10,11], "Emp Name": ["Olivia", "Liam", "Emma", "Noah", "Nathan", "Brooke"], "Department": ["HR", "IT", "IT", "Finance","Sales", "HR"]}
pd.concat([pd.DataFrame(D1), pd.DataFrame(D2)], axis = 0, ignore_index = True)
#concatenate the two DataFrames vertically (stacking them on top of each other) and reset the index of the resulting DataFrame. The "ignore_index=True" parameter ensures that the index
#axis=0 means we are concatenating along the rows (vertically). This will combine the two DataFrames into a single DataFrame with all the rows from both D1 and D2, and the index will be reset to a default integer index.
#The resulting DataFrame will contain all the rows from both D1 and D2, with the index starting from 0 and incrementing by 1 for each row.
```

```python
import pandas as pd
data = {"name": ["Alice", "Roger", "Dean", "Garret", "Hannah"], "age": [26, 43, 28, 28, 19], "profession": ["Engineer", "CEO", "Hockey Player", "Hockey Player", "Singer"]}

df = pd.DataFrame(data)
df[["name", "profession"]].applymap(lambda x: len(x)) 
#applymap() is a method in pandas that allows you to apply a function to each element of a DataFrame. In this case, the lambda function is used to calculate the length of each string in the "name" and "profession" columns. The result will be a new DataFrame where each value represents the length of the corresponding string from the original DataFrame.


df[["name", "profession"]].apply(lambda x: len(x), axis = 0) 
df[["name", "profession"]].apply(lambda x: len(x), axis = 1) 
```
